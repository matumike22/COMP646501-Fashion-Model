# Fashion-MNIST training on Colab

Select **Runtime → Change runtime type → GPU**, then run the cells in order.
The notebook uses the same training entry point as the terminal. Each run cross-validates, selects an epoch, and trains a fresh model on all training images before exporting the checkpoint.

```python
from pathlib import Path
import subprocess
import sys

repo = Path("/content/coursework")
repo_url = "https://github.com/matumike22/COMP646501-Fashion-Model.git"
repo_ref = "main"

if not repo.exists():
    subprocess.run(["git", "clone", "--branch", repo_ref, repo_url, str(repo)], check=True)
else:
    subprocess.run(["git", "pull", "--ff-only", "origin", repo_ref], cwd=repo, check=True)

subprocess.run(["git", "rev-parse", "--short", "HEAD"], cwd=repo, check=True)
```

```python
# Set to True to repeat the training-only horizontal flip experiment.
horizontal_flip = False

command = [sys.executable, "-u", "-m", "submission.fashion_training", "--gpu"]
if horizontal_flip:
    command.append("--horizontal-flip")

subprocess.run(command, cwd=repo, check=True)
```
